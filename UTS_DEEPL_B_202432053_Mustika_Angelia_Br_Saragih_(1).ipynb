  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432053_Mustika Angelia Br Saragih
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432053_Mustika Angelia Br Saragih
# 1. Upload dataset CSV langsung (akan muncul tombol 'Choose Files' di bawah sel ini)
import io
try:
    from google.colab import files
    uploaded = files.upload()                      # pilih file PRDECT-ID Dataset.csv
    nama_file = next(iter(uploaded))
    df = pd.read_csv(io.BytesIO(uploaded[nama_file]))
    print('File dimuat:', nama_file)
except ImportError:
    # Jika dijalankan di luar Colab (Jupyter lokal), baca dari folder yang sama
    df = pd.read_csv('PRDECT-ID Dataset.csv')
print('Ukuran awal :', df.shape)
print(df['Sentiment'].value_counts())

# 2. Pembersihan data & pra-pemrosesan NLP
df = df[['Customer Review', 'Sentiment']].dropna().drop_duplicates().reset_index(drop=True)
print('Setelah dedup:', df.shape)

def clean_text(text):
    text = str(text).lower()                       # case folding
    text = re.sub(r'http\S+|www\.\S+', ' ', text)   # hapus URL
    text = re.sub(r'[^a-z\s]', ' ', text)          # hapus angka, tanda baca, emoji
    text = re.sub(r'\s+', ' ', text).strip()        # rapikan spasi
    return text

df['clean'] = df['Customer Review'].apply(clean_text)
df = df[df['clean'].str.len() > 0].reset_index(drop=True)
df['label'] = (df['Sentiment'] == 'Positive').astype(int)   # Positive=1, Negative=0
df['tokens'] = df['clean'].str.split()                      # tokenisasi

# 3. Split train/test (80:20, stratified)
train_df, test_df = train_test_split(df, test_size=0.2, random_state=SEED, stratify=df['label'])
print('Train:', len(train_df), '| Test:', len(test_df))

# 4. Vocabulary mapping (hanya dari data latih agar tidak data leakage)
MIN_FREQ = 2
counter = Counter(tok for toks in train_df['tokens'] for tok in toks)
vocab = {'<PAD>': 0, '<UNK>': 1}
for word, freq in counter.items():
    if freq >= MIN_FREQ:
        vocab[word] = len(vocab)
vocab_size = len(vocab)
print('Ukuran vocabulary:', vocab_size)

# 5. Encoding + padding (pre-padding agar token terakhir LSTM = kata asli)
MAX_LEN = 50
def encode(tokens):
    ids = [vocab.get(t, 1) for t in tokens][:MAX_LEN]
    return [0] * (MAX_LEN - len(ids)) + ids

X_train = torch.tensor([encode(t) for t in train_df['tokens']], dtype=torch.long)
y_train = torch.tensor(train_df['label'].values, dtype=torch.float32)
X_test  = torch.tensor([encode(t) for t in test_df['tokens']], dtype=torch.long)
y_test  = torch.tensor(test_df['label'].values, dtype=torch.float32)
print('X_train:', X_train.shape, '| X_test:', X_test.shape)

# 6. DataLoader
BATCH_SIZE = 64
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(TensorDataset(X_test, y_test), batch_size=BATCH_SIZE, shuffle=False)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432053_Mustika Angelia Br Saragih
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=64, hidden_size=64, num_layers=1, dropout=0.3):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, hidden_size, num_layers=num_layers, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        emb = self.dropout(self.embedding(x))      # (batch, seq_len, embed_dim)
        _, (h_n, _) = self.lstm(emb)               # h_n: (num_layers, batch, hidden)
        out = self.dropout(h_n[-1])                # hidden state terakhir layer teratas
        return self.fc(out).squeeze(1)             # logit (batch,)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432053_Mustika Angelia Br Saragih
# 1. Hyperparameter
epochs = 20
learning_rate = 0.001
embed_dim = 64
hidden_size = 64

# 2. Model, loss, optimizer
model = SequenceModel(vocab_size, embed_dim, hidden_size).to(device)
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)

# 3. Riwayat
train_losses, test_losses = [], []
train_accs, test_accs = [], []

def evaluate(loader):
    model.eval()
    total_loss, correct, n = 0.0, 0, 0
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb)
            total_loss += criterion(logits, yb).item() * xb.size(0)
            correct += ((torch.sigmoid(logits) >= 0.5).float() == yb).sum().item()
            n += xb.size(0)
    return total_loss / n, correct / n

# 4. Training loop
for epoch in range(epochs):
    model.train()
    running_loss, correct, n = 0.0, 0, 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        running_loss += loss.item() * xb.size(0)
        correct += ((torch.sigmoid(logits) >= 0.5).float() == yb).sum().item()
        n += xb.size(0)
    tr_loss, tr_acc = running_loss / n, correct / n
    te_loss, te_acc = evaluate(test_loader)
    train_losses.append(tr_loss); train_accs.append(tr_acc)
    test_losses.append(te_loss);  test_accs.append(te_acc)
    print(f'Epoch {epoch+1:02d}/{epochs} | Train Loss {tr_loss:.4f} Acc {tr_acc:.4f} | Test Loss {te_loss:.4f} Acc {te_acc:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432053_Mustika Angelia Br Saragih
# 1. Grafik Loss & Accuracy terhadap Epoch
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
ax[0].plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
ax[0].plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
ax[0].set_title('Grafik Penurunan Loss Selama Pelatihan')
ax[0].set_xlabel('Epoch'); ax[0].set_ylabel('Loss')
ax[0].grid(True, linestyle='--', alpha=0.5); ax[0].legend()
ax[1].plot(train_accs, color='#114F6A', linewidth=2, label='Training Accuracy')
ax[1].plot(test_accs, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Accuracy')
ax[1].set_title('Grafik Accuracy Selama Pelatihan')
ax[1].set_xlabel('Epoch'); ax[1].set_ylabel('Accuracy')
ax[1].grid(True, linestyle='--', alpha=0.5); ax[1].legend()
plt.tight_layout(); plt.show()

# 2. Prediksi pada data uji
model.eval()
with torch.no_grad():
    probs = torch.sigmoid(model(X_test.to(device))).cpu().numpy()   # skor sentimen (prob. Positive)
preds = (probs >= 0.5).astype(int)
actual = y_test.numpy().astype(int)

print('Accuracy:', round(accuracy_score(actual, preds), 4))
print(classification_report(actual, preds, target_names=['Negative', 'Positive']))

# 3. Confusion matrix + distribusi skor prediksi
cm = confusion_matrix(actual, preds)
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
im = ax[0].imshow(cm, cmap='GnBu')
ax[0].set_xticks([0, 1]); ax[0].set_yticks([0, 1])
ax[0].set_xticklabels(['Negative', 'Positive']); ax[0].set_yticklabels(['Negative', 'Positive'])
ax[0].set_xlabel('Prediksi'); ax[0].set_ylabel('Aktual'); ax[0].set_title('Confusion Matrix Data Uji')
for i in range(2):
    for j in range(2):
        ax[0].text(j, i, cm[i, j], ha='center', va='center', fontsize=14, fontweight='bold')
plt.colorbar(im, ax=ax[0])

ax[1].hist(probs[actual == 0], bins=20, alpha=0.7, color='#114F6A', label='Aktual Negative')
ax[1].hist(probs[actual == 1], bins=20, alpha=0.7, color='#1FA4B1', label='Aktual Positive')
ax[1].axvline(0.5, color='red', linestyle='--', label='Threshold 0.5')
ax[1].set_title('Distribusi Skor Prediksi Sentimen')
ax[1].set_xlabel('Skor (Probabilitas Positive)'); ax[1].set_ylabel('Jumlah Ulasan')
ax[1].legend(); ax[1].grid(True, linestyle='--', alpha=0.5)
plt.tight_layout(); plt.show()

# 4. Contoh hasil prediksi beserta skornya
sample = test_df.reset_index(drop=True).head(10).copy()
sample['Aktual'] = np.where(actual[:10] == 1, 'Positive', 'Negative')
sample['Prediksi'] = np.where(preds[:10] == 1, 'Positive', 'Negative')
sample['Skor'] = probs[:10].round(4)
sample['Review'] = sample['Customer Review'].str.slice(0, 70)
display(sample[['Review', 'Aktual', 'Prediksi', 'Skor']])
```
